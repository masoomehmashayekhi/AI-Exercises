# ۱ – آماده‌سازی داده (Persian News Dataset)

در این نوت‌بوک، دیتاست خبری فارسی را از پوشه‌ی `data/raw` می‌خوانید، یک زیرمجموعه‌ی کوچک و قابل‌مدیریت از آن می‌سازید، داده‌ها را کمی تمیز می‌کنید و نتیجه را در مسیر `data/processed/news_subset.csv` ذخیره می‌کنید.

        https://www.kaggle.com/datasets/amirzenoozi/persian-news-dataset/

</div>


```python
# TODO: import های لازم را این‌جا بنویسید
# مثال:
import pandas as pd
from pathlib import Path
```

```python
# TODO: مسیر فایل خام (CSV) و مسیر خروجی را تنظیم کنید
raw_path = Path('../data/raw/archive_v5.csv')
subset_path = Path('../data/processed/news_subset.csv')
```

```python
# TODO: دیتاست را با pandas بخوانید و چند سطر اول را نمایش دهید
df = pd.read_csv(raw_path)

print(df.shape)
df.head()
```

```python
# TODO: ردیف‌های نامناسب (مثلاً متن خالی یا خیلی کوتاه) را حذف کنید
df = df.dropna()
df = df[df["body"].str.len() > 200]
df["text"] = df["title"].fillna("") + "\n\n" + df["body"].fillna("")
print(df.shape)

text_cols = ["title", "abstract", "body","published_datetime","service","text"]

df[text_cols] = (
    df[text_cols]
    .fillna("")
    .astype(str)
)
print(df.shape)
```

```python
# TODO: یک زیرمجموعه‌ی کوچک بسازید (مثلاً چند Category یا نمونه‌گیری تصادفی)
df_subset = df.sample(n=2000, random_state=42)
df_subset = df_subset.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
# TODO: یک تابع ساده برای نرمال‌سازی حروف فارسی (ی/ك) بنویسید و روی متن اعمال کنید
def clean_persian_text(text):
    if not isinstance(text, str):
        return ""
    text = text.replace("ي", "ی").replace("ك", "ک")
    text = text.replace("\u200c", " ")
    text = text.replace("\n", " ").strip()
    return text

df_subset["text"] = df_subset["text"].apply(clean_persian_text)
```

```python
# TODO: دیتافریم نهایی را در subset_path به صورت CSV ذخیره کنید
df_subset.to_csv(subset_path, index=False, encoding="utf-8-sig")

print("Saved:", subset_path)
print(df_subset.shape)
```
